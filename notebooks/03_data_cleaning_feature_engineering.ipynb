# Data Cleaning & Feature Engineering

## Objective
Prepare the banking customer churn dataset for SQL analysis, machine learning, and business intelligence.

## Tasks
- Validate data types and missing values
- Remove duplicate records if necessary
- Handle categorical variables
- Create meaningful customer segments
- Create features that may improve churn analysis
- Validate the processed dataset
- Export the cleaned dataset for downstream analysis

```python
import pandas as pd
import numpy as np

# Load the raw dataset
df = pd.read_csv(
    r"C:\Users\amish\OneDrive\Desktop\portfolio\fintech-customer-churn-retention\data\raw\Bank Customer Churn Prediction.csv"
)

# Check the dataset
print("Dataset shape:", df.shape)
df.head()
```

```python
# Check missing values
print("Missing values by column:")
print(df.isnull().sum())

# Check duplicate rows
print("\nNumber of duplicate rows:", df.duplicated().sum())
```

```python
# Check data types
print("Data types:")
print(df.dtypes)

# Check unique values for categorical/binary columns
print("\nCountry values:")
print(df["country"].unique())

print("\nGender values:")
print(df["gender"].unique())

print("\nCredit card values:")
print(df["credit_card"].unique())

print("\nActive member values:")
print(df["active_member"].unique())

print("\nChurn values:")
print(df["churn"].unique())
```

```python
# Check ranges of numerical variables
print("Credit score:", df["credit_score"].min(), "to", df["credit_score"].max())
print("Age:", df["age"].min(), "to", df["age"].max())
print("Tenure:", df["tenure"].min(), "to", df["tenure"].max())
print("Balance:", df["balance"].min(), "to", df["balance"].max())
print("Products:", df["products_number"].min(), "to", df["products_number"].max())
print("Estimated salary:", df["estimated_salary"].min(), "to", df["estimated_salary"].max())

# Check binary variables
print("\nCredit card:", sorted(df["credit_card"].unique()))
print("Active member:", sorted(df["active_member"].unique()))
print("Churn:", sorted(df["churn"].unique()))
```

```python
# Create a working copy of the raw dataset
clean_df = df.copy()

# Confirm the copy
print("Original dataset shape:", df.shape)
print("Clean dataset shape:", clean_df.shape)
```

```python
# Create age groups
clean_df["age_group"] = pd.cut(
    clean_df["age"],
    bins=[17, 25, 35, 45, 55, np.inf],
    labels=["18-25", "26-35", "36-45", "46-55", "56+"]
)

# Check the result
print(clean_df["age_group"].value_counts().sort_index())
```

```python
# Create balance groups
clean_df["balance_group"] = pd.cut(
    clean_df["balance"],
    bins=[-1, 50000, 100000, 150000, 200000, np.inf],
    labels=["0-50K", "50K-100K", "100K-150K", "150K-200K", "200K+"]
)

# Check the result
print(clean_df["balance_group"].value_counts().sort_index())
```

```python
# Create credit score groups
clean_df["credit_score_group"] = pd.cut(
    clean_df["credit_score"],
    bins=[349, 579, 669, 739, 799, 850],
    labels=["Poor", "Fair", "Good", "Very Good", "Excellent"]
)

# Check the result
print(clean_df["credit_score_group"].value_counts().sort_index())
```

```python
# Create estimated salary groups
clean_df["salary_group"] = pd.cut(
    clean_df["estimated_salary"],
    bins=[-1, 50000, 100000, 150000, np.inf],
    labels=["0-50K", "50K-100K", "100K-150K", "150K+"]
)

# Check the result
print(clean_df["salary_group"].value_counts().sort_index())
```

```python
# Create customer activity segment
clean_df["activity_segment"] = np.where(
    clean_df["active_member"] == 1,
    "Active",
    "Inactive"
)

# Check the result
print(clean_df["activity_segment"].value_counts())
```

```python
# Create product segment
clean_df["product_segment"] = pd.cut(
    clean_df["products_number"],
    bins=[0, 1, 2, 4],
    labels=["1 Product", "2 Products", "3-4 Products"]
)

# Check the result
print(clean_df["product_segment"].value_counts().sort_index())
```

```python
# Check missing values in engineered features
engineered_features = [
    "age_group",
    "balance_group",
    "credit_score_group",
    "salary_group",
    "activity_segment",
    "product_segment"
]

print(clean_df[engineered_features].isnull().sum())
```

```python
# Final validation of the cleaned dataset

print("Original shape:", df.shape)
print("Cleaned shape:", clean_df.shape)

print("\nCleaned dataset columns:")
print(clean_df.columns.tolist())

print("\nMissing values:")
print(clean_df.isnull().sum().sum())

print("\nDuplicate rows:")
print(clean_df.duplicated().sum())
```

```python
# Final quality check for engineered features

for col in engineered_features:
    print(f"\n{col}:")
    print(clean_df[col].unique())
```

```python
import os

# Create processed data folder
processed_path = r"C:\Users\amish\OneDrive\Desktop\portfolio\fintech-customer-churn-retention\data\processed"

os.makedirs(processed_path, exist_ok=True)

print("Processed data folder is ready.")
```

```python
# Export cleaned and feature-engineered dataset

output_file = os.path.join(
    processed_path,
    "bank_customer_churn_processed.csv"
)

clean_df.to_csv(output_file, index=False)

print("Processed dataset saved successfully!")
print(output_file)
```

```python
# Verify the exported processed dataset

processed_df = pd.read_csv(output_file)

print("Processed dataset shape:", processed_df.shape)
print("\nMissing values:", processed_df.isnull().sum().sum())
print("Duplicate rows:", processed_df.duplicated().sum())

print("\nFirst 5 rows:")
display(processed_df.head())
```
